```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# 1. Load the total dataset
df = pd.read_csv('../data/raw/apachejit_total.csv')

print("="*40)
print(f"Total Commits: {df.shape[0]:,}")
print(f"Total Features: {df.shape[1]}")
print("="*40)

# 2. Confirm Columns
print("\n[ Column Names ]")
print(df.columns.tolist())

# 3. Class Balance (The 'fix' flag indicates a bug-inducing commit)
print("\n[ Class Balance (Bug-Inducing vs Clean) ]")
print(df['fix'].value_counts())
print(df['fix'].value_counts(normalize=True).round(4) * 100, "%")

# 4. Project Distribution
print("\n[ Commits per Project ]")
if 'project' in df.columns:
    print(df['project'].value_counts())

# 5. Duplicates Check
duplicates = df.duplicated().sum()
print(f"\n[ Duplicate Rows ]: {duplicates}")

# 6. Visualize Feature Skewness (Lines Added 'la' & Lines Deleted 'ld')
# Software engineering metrics follow a power-law (most commits are tiny, a few are massive).
# We apply np.log1p() to squash the massive tail so the ML model can digest it.
skewed_features = ['la', 'ld', 'nf'] 

fig, axes = plt.subplots(len(skewed_features), 2, figsize=(12, 4 * len(skewed_features)))

for i, col in enumerate(skewed_features):
    if col in df.columns:
        # Raw Data Plot
        df[col].hist(bins=50, ax=axes[i, 0], color='#3498db', edgecolor='black', alpha=0.7)
        axes[i, 0].set_title(f"Raw '{col}' (Highly Skewed)")
        axes[i, 0].set_yscale('log') # Log scale on Y just to make the tail visible
        
        # Log-Transformed Data Plot
        np.log1p(df[col]).hist(bins=50, ax=axes[i, 1], color='#e74c3c', edgecolor='black', alpha=0.7)
        axes[i, 1].set_title(f"Log1p Transformed '{col}' (Ready for ML)")

plt.tight_layout()
plt.show()
```
